```python
from glob import glob
import pandas as pd
import numpy as np

def check_progress(outdir, exp_id, indexes_all):
    files_suc_samples = glob(f'{outdir}/{exp_id}/progress_prc_*')
    if len(files_suc_samples) != 0:
        dfs = []
        for i in files_suc_samples:
            df = pd.read_csv(i)
            dfs.append(df)
    final_suc_samples = pd.concat(dfs, ignore_index=True)
    indexes_to_exclude = final_suc_samples['sample_idx'].values
    indexes_remaining = np.setdiff1d(indexes_all, indexes_to_exclude)
    return len(indexes_remaining)
```

```python
import os
from pathlib import Path

indexes_all = np.arange(200)
outdir = '../experiments_sonar_v0'

files_suc_samples = [x.name for x in Path(outdir).iterdir() if x.is_dir()]

for f in files_suc_samples:
    print(f'{f} \t {check_progress(outdir, f, indexes_all)}')
```

```python
import os, re
from glob import glob
import pandas as pd
import numpy as np

def check_remaining_samples(exp_dir, exp_id, n_samples=None):
    if n_samples is None:
        if 'llmseg' in exp_id or 'refcocog' in exp_id:
            n_samples = 500
        else:
            n_samples = 779
    files_suc_samples = glob(f'{exp_dir}/{exp_id}/progress_prc_*')
    if len(files_suc_samples) != 0:
        dfs = []
        for i in files_suc_samples:
            df = pd.read_csv(i)
            dfs.append(df)
        final_suc_samples = pd.concat(dfs, ignore_index=True)
        indexes_to_exclude = final_suc_samples['sample_idx'].values
        indexes_remaining = np.setdiff1d(np.arange(n_samples), indexes_to_exclude)
        return len(indexes_remaining)
    return n_samples


folder_path = '/home/jovyan/zinkovich/ref-seg-text-break/experiments/EMNLP-paper/sonar'

datasets = {
    'reason_test': 779,
    'refcocog_test': 1000,
    'llmseg_test': 1000,
}

models = [
    'lisa-7b-v1',
    'lisa-7b-v1-exp',
    'lisa-13b-v0',
    'lisa-13b-v0-exp',
    'lisa-13b-v1',
    'lisa-13b-v1-exp',
    'lisa++',
    'gsva-7b-ft-res',
    'gsva-13b-ft-res',
    'gsva-13b-llama2-ft-res',
]

for dataset, max_samples in datasets.items():
    for model in models:
        path = f'{folder_path}/{dataset}'
        n_ramin = check_remaining_samples(path, model, max_samples)
        print(f'{dataset} {model:20} \t left = {n_ramin}')
    print('\n')
```

```python
text_with_system_prompt = '\n In agricultural management, it is necessary to spray pesticides on crops in a timely manner to prevent various diseases. What in the picture could be threatened during this process? Please output segmentation mask.'
```

```python
start_adv_str = text_with_system_prompt.find('\n ')
end_adv_str = text_with_system_prompt.find('Please output segmentation mask.')

start_adv_str += len('\n ')
```

```python
text_with_system_prompt[start_adv_str:end_adv_str]
```

```python
max_word_length = 15
max_sublength = 10
stop_words = ['.', ',', '?', '!']
```

```python
if len(text_with_system_prompt[start_adv_str:end_adv_str].split()) > max_word_length:
    for i, letter in enumerate(text_with_system_prompt[start_adv_str:end_adv_str]):
        if letter in stop_words and \
            len(text_with_system_prompt[start_adv_str:start_adv_str+i].split()) > max_sublength and \
            start_adv_str + i + 2 < end_adv_str:
            end_adv_str = start_adv_str + i + 1
            break
```

```python
import transformers

tokenizer = transformers.AutoTokenizer.from_pretrained(
    pretrained_model_name_or_path='xinlai/LISA-13B-llama2-v1-explanatory',
    model_max_length=1024,
    padding_side='right',
    use_fast=True,
    legacy=False,
)
tokenizer.pad_token = tokenizer.unk_token
```

```python
from lisa.utils.dataset import ValDataset

dataset = ValDataset(
    base_image_dir='../dataset/llm-seg40k',
    image_size=1024,
    val_dataset='llm|seg|40k|val',
    vision_tower="openai/clip-vit-large-patch14",
    tokenizer=tokenizer,
)
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
import cv2

n = 162
image_path = dataset[n][0]
img = Image.open(image_path)

print(dataset[n][3][0])
pred_mask = dataset[n][4][0].numpy().astype(np.uint8)

plt.figure(figsize=(10, 10))
color = 'Oranges'
line_color = 'orange'
plt.imshow(np.ma.masked_where(pred_mask == 0, pred_mask), cmap=color, interpolation='none', alpha=0.3)
contours, _ = cv2.findContours(pred_mask.astype(np.uint8), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
for contour in contours:
    plt.plot(contour[:, 0, 0], contour[:, 0, 1], color=line_color, linewidth=4)
plt.imshow(img)
plt.axis('off')
plt.savefig('tmp.png', bbox_inches='tight', pad_inches=0, dpi=300)
plt.show() 
```
